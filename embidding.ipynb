```python
import pandas as pd
import numpy as np
import tensorflow as tp

```

```python
sentences = [
    "Python is a powerful programming language.",
    "I love solving problems using Python.",
    "Machine learning is fun to learn.",
    "FastAPI is great for building APIs.",
    "Django provides an excellent admin interface.",
    "Data structures make programs efficient.",
    "Artificial Intelligence is changing the world.",
    "I am learning how to create chatbots.",
    "Lists and dictionaries are core Python data types.",
    "Practice makes you a better programmer."
]
```

```python
## define the vocabulary size
voc_size = 10000
```

```python
## one hot representation
embidded_words = [tp.keras.preprocessing.text.one_hot(word,voc_size) for word in sentences]
```

```python
embidded_words
```

```python
## word embidding representation
sent_len = 8
embidded_doc = tp.keras.utils.pad_sequences(embidded_words,padding='pre',maxlen=sent_len)
embidded_doc
```

```python
# feature representation
dim = 10
```

```python
model = tp.keras.models.Sequential()
```

```python
model.add(tp.keras.layers.Embedding(voc_size,dim,input_length=sent_len))
model.compile('adam','mse')
model.build((None, sent_len))
```

```python
model.summary()
```

```python
model.predict(embidded_doc)
```
